### Строки

**Строка** (string, тип `str`) — это упорядоченная последовательность символов, предназначенная для хранения и обработки текстовой информации.

В Python строки являются **неизменяемыми** (immutable) объектами: после создания строку нельзя изменить, любые операции "изменения" создают новую строку.

```python
s = "Привет, мир!"
print(type(s))
```

### Способы создания строк

1 - в кавычках прописать текст
(кавычки одинарные, двойные, тройные - для многострочного текста)

```python
s1 = 'Python'
s2 = "Python"
s3 = """Многострочная
строка"""
s4 = '''Тоже многострочная'''
```

2 пустой объект строки

```python
text = str()
print(type(text))
```

3 - генерация путем повторения

```python
text = '-' * 100
print(text)
```

### Спецсимволы внутри строки

Иногда необходимо оформить текст внутри строки - добавить переходы на нову строку, табулирование и тд:

```python
s7 = "Первая строка\nВторая строка"  # \n — перевод строки
s8 = "Колонка1\tКолонка2"           # \t — табуляция
s9 = "Обратный слэш: \""            # \\ — сам слэш

print(s7)
print(s8)
print(s9)
```

### Кодировки строк

узнать технические характеристики можно с помощью встроенных функций?

Длина строки - количество символов - даже пробельных

```python
text = "some text for example                   "
print(len(text))
```

Код символа

```python
code = ord('☼')
print(code)
```

символ по коду

```python
print(chr(49))
```

преобразование строки в байты

```python
bytes_text = text.encode()
print(bytes_text)

for c in bytes_text:
    print(c)
```
