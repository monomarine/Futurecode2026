Строка — это последовательность, к символам можно обращаться по индексу (с нуля). Отрицательные индексы считаются с конца.

```python
s = "Python"

print(s[0])    # 'P'
print(s[5])    # 'n'
print(s[-1])   # 'n' — последний
print(s[-6])   # 'P' — первый

# Срез: s[start:stop:step] (stop не включается)
print(s[0:3])   # 'Pyt'
print(s[2:])    # 'thon'
print(s[:3])    # 'Pyt'
print(s[:])     # 'Python' — копия
print(s[::2])   # 'Pto' — каждый второй
print(s[::-1])  # 'nohtyP' — переворот строки
print(s[-3:])   # 'hon'
```
