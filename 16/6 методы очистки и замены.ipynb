### обрезка

strip = обрезка с обоих сторон
lstrip - с левой стороны
rstrip - с правой стороны

```python
from shlex import join

text = "   hello!   "
print(text.strip())
print(text.lstrip())
print(text.rstrip())
```

```python
text = "...somedata!!!"
print(text.strip('!'))
print(text.strip('!.?'))
```

### Замена текста

```python
s1 = "hello world".replace("o", "0")        # 'hell0 w0rld'
s2 = "hello world".replace("l", "|", 2)

print(s1)
print(s2)
```

removeprefix / removesuffix (Python 3.9+)

```python
s1 = "unhappy".removeprefix("un")   # 'happy'
s2 = "file.txt".removesuffix(".txt")  # 'file'

print(s1)
print(s2)
```

### Таблицы перевода

Замена символов по таблице (работает быстрее replace при множестве замен)

```python
table = str.maketrans("abc", "xyz")
res = "cab____abc".translate(table)   # 'xyzxyz'
print(res)

# Удаление символов (третий аргумент — что удалить)
table = str.maketrans("", "", "aeiou")
"hello world".translate(table)  # 'hll wrld'
```

### Разбиение и объединение

**split** - разделяет (режет) исходную последовательность

```python
text = "я изучаю питон. в нем а  много интересного"

print(text.split())
print(text.split('.'))
print(text.split('а'))
```

`split()` без аргумента схлопывает подряд идущие пробелы и убирает пустые строки, а `split(" ")` — нет:

```python
"a  b".split()      # ['a', 'b']
"a  b".split(" ")   # ['a', '', 'b']
```

**join** - объединяет уже нарезанную последовательность обратно в строку

join применяется к разделителю (символу или тексту)

```python
s1 = ", ".join(["a", "b", "c"])
s2 = "".join(["P", "y"])
s3 = "-".join("abc")

print(s1)
print(s2)
print(s3)
```

```python
data = ["some", "text"]
print(' '.join(data))
```

**partition / rpartition** - делит строку ровно на 3 части: (до, разделитель, после) — разделитель сохраняется в результате.

всегда возвращают кортеж из 3 строк — даже если разделитель не найден.

```python
s = "user@example.com"

print(s.split("@"))
print(s.partition("@"))
```

### выравнивание и заполнение строк

```python
s1 = "hi".center(10, "*")   # '****hi****'
s2 = "hi".ljust(10, ".")    # 'hi........'
s3 = "hi".rjust(10, ".")    # '........hi'
s4 = "42".zfill(5)          # '00042' — заполнение нулями слева

print(s1)
print(s2)
print(s3)
print(s4)
```
