### перевод в верхний регистр (Upper)

```python
text = 'hello world'
text = text.upper()

print(text)
```

### Перевод в нижний регистр (lower)

```python
text = "ПроИзВолЬный РЕГИСтр"
text = text.lower()

print(text)
```

### Первый символ заглавный- остальные строчные

```python
text = text.capitalize()
print(text)
```

### Каждое слово с заглавной буквы

```python
text = text.title()
print(text)
```

### Инверсия регистра

```python
text = text.swapcase()
print(text)
```
