Все перечисленные ниже метожы возращают Bool - результат проверки

```python
text = 'простой текст для анализа 1990 - 2026. #new'
```

#### в строке только буквы и цифры

```python
print(text.isalnum())
```

#### в строке только буквы

```python
print(text.isalpha())
```

#### в строке только цифры

```python
print(text.isdigit())
```

#### в строке только десятичные цифры

```python
"462814816".isdecimal()
```

#### проверка на регистр

```python
print(text.islower())
print(text.isupper())
```

#### каждое слово с заглавной

```python
print(text.istitle())
```
