- find() - первое вхождение подстроки при чтении слева направо
- rfind() - первое вхождение подстроки при чтении справа налево

аналогично:
- index()
- rindex()

**Разница:** `find` возвращает `-1`, `index` бросает `ValueError`.

```python
text = "арозаупаланалапуазора"

print(text.find('роза'))
print(text.index('роза'))

print(text.rfind('роз'))
print(text.rfind('abc'))
```

```python
print(text.find('abc'))
print(text.index('abc'))
```
