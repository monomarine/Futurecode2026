## Логические операторы в условиях

Условие может объединять несколько проверок через `and`, `or`, `not`.

```python
age = 18
has_passport = True

if age >= 18 and has_passport:
    print("Можно купить билет")
```

```python
x = 15

# Диапазон: 10 < x < 20
if x > 10 and x < 20:
    print("в диапазоне")

# Цепочка сравнений (питоновский способ, то же самое)
if 10 < x < 20:
    print("в диапазоне")

# Или
if x < 0 or x > 100:
    print("вне диапазона")

# Отрицание
if not (x == 0):
    print("не ноль")
```

### Проверка на `None`

`None` — специальное значение «ничего». Проверять нужно через `is`, а не через `==`:


```python
result = None

if result is None:
    print("Результата нет")
```

### Проверка вхождения

```python
fruits = ["яблоко", "банан", "апельсин"]

if "банан" in fruits:
    print("Банан есть")

if "киви" not in fruits:
    print("Киви нет")

text = "Python — это просто"
if "Python" in text:
    print("Упоминается Python")
```
