## Условие

**Условие** — это выражение, которое может быть **истинным** (`True`) или **ложным** (`False`). От результата условия зависит, какой блок кода выполнится.

Условия — основа «ветвления» программы. Без них программа могла бы только последовательно выполнять команды сверху вниз. С условиями программа может **выбирать** путь в зависимости от данных.

**Аналогия из жизни:**

> Если на улице дождь — взять зонт. Иначе — не брать.

В программе:

```python
if идет_дождь:
    взять_зонт()
else:
    не_брать_зонт()
```

**Что может быть условием:**

- результат сравнения: `x > 5`, `name == "Анна"`
- логическое значение: `True`, `False`
- переменная типа `bool`: `is_student`
- любое значение, приводимое к `bool`
- логическое выражение: `x > 0 and y < 10`
- результат вызова функции, возвращающей `bool`

````python
age = 18

if age >= 18:              # сравнени
    print("Совершеннолетний")

if True:                    # литерал
    print("Всегда выполнится")

items = [1, 2, 3]
if items:                   # непустой список → True
    print("Список не пуст")
````

 ## Базовый синтаксис `if` / `else`

### Простейший `if`

```python
if условие:
    # блок кода, который выполнится, если условие истинно
    действие
```

**Синтаксические правила (обязательны!):**

1. После `if` — условие, затем **двоеточие `:`**.
2. Тело `if` — **с отступом** (обычно 4 пробела).
3. Отступ определяет, что входит в блок.

````python
x = 10

if x > 5:
    print("x больше пяти")
    print("Это тоже часть блока")
print("А это — уже вне блока, выполнится всегда")
````

**Если условие ложно** — блок пропускается целиком:

````python
x = 3

if x > 5:
    print("x больше пяти")   # не выполнится
print("Программа продолжается")
````

### Отступы — это синтаксис, а не стиль

В отличие от C/C++/Java, где блоки выделяются фигурными скобками `{}`, в Python **блок определяется отступом**. Это часть грамматики языка, а не рекомендация.

````python
# Правильно
if x > 5:
    print("раз")
    print("два")

# Ошибка: IndentationError
if x > 5:
print("раз")
````

**Правила отступов:**

- Все строки одного блока должны иметь **одинаковый отступ**.
- Обычно используется **4 пробела** (PEP 8). Не смешивайте табы и пробелы.
- Вложенные блоки получают дополнительный отступ.

````python
x = 15
if x > 5:
    print("внешний блок")
    if x > 10:              # вложенный if
        print("вложенный блок")  # ещё больший отступ
    print("снова внешний блок")
````

### `if` / `else`

`else` выполняется, когда условие `if` **ложно**. `else` **не имеет своего условия** — он «по умолчанию».

```python
if условие:
    # если True
else:
    # если False
```

````python
age = int(input("Возраст: "))

if age >= 18:
    print("Доступ разрешён")
else:
    print("Доступ запрещён")
````

выполняется **ровно один** из двух блоков — либо `if`, либо `else`, но никогда оба.

### Тело из одной строки (не рекомендуется, но допускается)

Если в блоке одна инструкция, её можно записать в той же строке:

````python
if x > 0: print("положительное")
````

Это допустимо, но **не рекомендуется** для читаемости. Лучше всегда писать с отступом.

### Пустой блок: `pass`

Иногда нужно «заглушить» блок — например, при разработке. Для этого есть `pass`:

````python
x = 7
if x > 0:
    pass    # TODO: дописать позже
else:
    print("не положительное")
````

## Дополнительный блок условия `elif`

Часто нужно проверить **несколько взаимоисключающих вариантов**. Можно было бы написать вложенные `if`/`else`, но это громоздко. `elif` (сокращение от «else if») позволяет выстроить **цепочку проверок**.

```python
if условие1:
    # блок 1
elif условие2:
    # блок 2
elif условие3:
    # блок 3
else:
    # блок по умолчанию
```

1. Проверяется `условие1`. Если истинно — выполняется блок 1, **все остальные проверки пропускаются**.
2. Если ложно — проверяется `условие2`. Если истинно — блок 2, остальные пропускаются.
3. И так далее.
4. Если **все** условия ложны — выполняется `else` (если он есть).
5. `else` **необязателен**.

**Ключевой момент:** выполняется **ровно один** блок из всей цепочки — первый, условие которого истинно.

````python
score = int(input("Баллы (0-100): "))

if score >= 90:
    print("Отлично")
elif score >= 75:
    print("Хорошо")
elif score >= 60:
    print("Удовлетворительно")
else:
    print("Неудовлетворительно")
````

### `elif` без `else`

`else` необязателен. Если ни одно условие не истинно и `else` нет — **ничего не выполняется**:

````python
x = 5

if x > 10:
    print("больше 10")
elif x > 7:
    print("больше 7")
# для x = 5 ничего не выведется
````

### Несколько независимых `if`

**Важно различать цепочку `elif` и несколько отдельных `if`.**

````python
# Цепочка elif — выполнится РОВНО ОДИН блок
x = 15
if x > 10:
    print("A")     # выполнится
elif x > 5:
    print("B")     # пропустится
elif x > 0:
    print("C")     # пропустится
````

````python
# Несколько независимых if — каждый проверяется отдельно
x = 15
if x > 10:
    print("A")     # выполнится
if x > 5:
    print("B")     # тоже выполнится!
if x > 0:
    print("C")     # и это тоже!
````

**Когда что использовать:**

- `elif` — когда варианты **взаимоисключающие** (число либо положительное, либо отрицательное, либо ноль).
- Независимые `if` — когда проверки **независимы** (проверить, что число чётное, и отдельно — что оно положительное).
