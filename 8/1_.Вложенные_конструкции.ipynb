## вложенные конструкции

**Вложенная конструкция** — это конструкция (условие или цикл), которая находится **внутри** другой конструкции того же или другого типа.

**Общая идея:** внутренняя конструкция выполняется **многократно** — по одному разу для каждого «шага» внешней.

**Аналогия из жизни:**

> Для каждого дня недели (внешний цикл):
>     Для каждого урока в этот день (внутренний цикл):
>         посетить пару

Внешний цикл — «дни», внутренний — «уроки». Для **каждого** дня проходим **все** уроки.

**Схема вложенности:**

```python
внешняя_конструкция:
    # тело внешней
    внутренняя_конструкция:
        # тело внутренней (выполняется много раз)
    # продолжение тела внешней
```

**Правило отступов:** каждый уровень вложенности добавляет отступ (обычно 4 пробела).

````python
for i in range(3):          # уровень 0
    for j in range(4):      # уровень 1 (отступ 4)
        print(i, j)         # уровень 2 (отступ 8)
````

## Вложенные условия

**Вложенное условие** — `if` внутри `if`. Внутреннее условие проверяется **только если** внешнее истинно.

```python
if условие1:
    # блок 1
    if условие2:
        # блок 1.1 (только если условие1 И условие2)
    else:
        # блок 1.2 (условие1 истинно, условие2 ложно)
else:
    # блок 2 (условие1 ложно; условие2 даже не проверяется)
```

````python
x = int(input("Число: "))

if x != 0:
    if x > 0:
        print("Положительное")
    else:
        print("Отрицательное")
else:
    print("Ноль")
````

внутренний `if x > 0` не проверяется, если `x == 0` — внешнее условие уже ложно.

### Вложенные условия в цикле
Вложенные условия часто встречаются **внутри циклов** — для фильтрации:

````python
numbers = [3, -1, 5, -2, 7, -8]

for x in numbers:
    if x > 0:
        if x % 2 == 0:
            print(f"{x} — положительное чётное")
        else:
            print(f"{x} — положительное нечётное")
    else:
        print(f"{x} — неположительное")
````

### Упрощение вложенных условий через `and`

````python
if x > 0:
    if x % 2 == 0:
        print("Положительное чётное")

# Через and (эквивалент)
if x > 0 and x % 2 == 0:
    print("Положительное чётное")
````

### Глубокая вложенность

````python
a = b = c = d = True

if a:
    if b:
        if c:
            if d:
                print("всё")
````

````python
if a and b and c and d:
    print("всё")
````
