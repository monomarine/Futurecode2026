
## Вложенные циклы

### Базовый синтаксис

```python
for i in range(внешний):
    for j in range(внутренний):
        # тело: выполняется внешний × внутренний раз
```

**Как работает:**

1. Берётся первое значение `i`.
2. Внутренний цикл **полностью** проходит все значения `j`.
3. Берётся второе значение `i` — и снова **все** `j`.
4. И так далее.

**Число итераций тела = (число итераций внешнего) × (число итераций внутреннего).**

````python
for i in range(3):
    for j in range(2):
        print(f"i={i}, j={j}")
````

**Всего итераций:** 3 × 2 = 6.

**Пошаговый разбор:**

| Шаг | `i` | `j` | Действие |
|-----|-----|-----|----------|
| 1 | 0 | 0 | print |
| 2 | 0 | 1 | print |
| — | 0 | — | внутренний цикл закончился, `i` = 1 |
| 3 | 1 | 0 | print |
| 4 | 1 | 1 | print |
| — | 1 | — | внутренний закончился, `i` = 2 |
| 5 | 2 | 0 | print |
| 6 | 2 | 1 | print |

**Ключевое:** внутренний цикл **начинается заново** для каждого значения `i`.

### Пример - таблица умножения

````python
for i in range(2, 10):
    for j in range(2, 10):
        print(f"{i} * {j} = {i * j}")
    print("---")
````

### Таблица Пифагора

````python
for i in range(1, 6):
    for j in range(1, 6):
        print(f"{i * j:5}", end=" ")
    print()
````

- `end=" "` — чтобы числа шли в одну строку.
- `print()` после внутреннего цикла — чтобы перейти на новую строку для следующего `i`.

### Прямоугольник из символов

````python
rows = 10
cols = 3

for i in range(rows):
    for j in range(cols):
        print("+", end=' ')
    print()
````

### Треугольник из символов

````python
n = 10
for i in range(1, n + 1):
    for j in range(i):
        print("+", end=" ")
    print()
````

### Равнобедренный треугольник из символов

````python
n = 5
for i in range(1, n + 1):
    # пробелы
    for j in range(n - i):
        print(" ", end=" ")
    # звёзды
    for j in range(2 * i - 1):
        print("+", end=" ")
    print()
````

- Пробелы: `n - i` (верхняя строка — максимум пробелов).
- Звёзды: `2*i - 1` (нечётное число).
