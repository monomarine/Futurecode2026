### Параметры и аргументы


| Характеристика | Параметры | Аргументы |
| :--- | :--- | :--- |
| **Что это?** | Переменные в определении функции | Фактические значения при вызове |
| **Где объявляются?** | В строке с `def` | В скобках при вызове функции |
| **Пример** | `def greet(name):` — здесь `name` параметр | `greet("Анна")` — здесь `"Анна"` аргумент |

### Функция с параметрами

```python
def greet(name):
    print(f"Привет, {name}!")

greet("Анна")  # "Анна" — аргумент
```

Порядок аргументов должен соответствовать порядку параметров.

```python
def divide(a, b):
    return a / b

def multiply(a, b):
    return a * b

result = multiply(3,5)
print(result)

print(divide(10, 2))   # 5.0 — верно
print(divide(2, 10))   # 0.2 — верно, но результат другой (порядок важен!)
```

### Именованные аргументы

Позволяют менять порядок передачи аргументов:

```python
print(divide(b=2, a=10))  # 5.0
print(divide(a = 200, b = 54))
```

### Параметры по умолчанию

Если аргумент не передан, используется значение по умолчанию:

```python
def pow(x, power=2):
    return x ** power

print(pow(3))      # 9 (power=2)
print(pow(3, 3))   # 27
```
