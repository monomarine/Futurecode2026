**Функция** — это именованный блок кода, который выполняет определенную задачу и может быть вызван многократно.

**Зачем нужны функции:**

- Устранение дублирования кода — не нужно писать один и тот же код несколько раз.
- Разбиение сложных задач на простые — код становится понятнее.
- Улучшение читаемости программы — код становится структурированным.
- Повторное использование — одну функцию можно вызывать в разных частях программы.

**Пример без функции (дублирование):**

````python
print("Привет, Анна!")
print("Привет, Иван!")
print("Привет, Мария!")
````

**Пример с функцией:**

````python
def greet(name):
    print(f"Привет, {name}!")

greet("Анна")
greet("Иван")
greet("Мария")
````

**Синтаксис создания функции**

Общий синтаксис:
```python
def имя_функции(параметры):
    """Документация (опционально)"""
    # тело функции
    return результат  # (опционально)
```
Компоненты:
- `def` — ключевое слово, с которого начинается объявление функции.
- `имя_функции` — обычно глагол или действие (например, calculate_sum, print_hello).
- `параметры` — переменные в круглых скобках, входные данные.
- `return` — возвращаемое значение (если нет return, функция возвращает None).

**Простая функция без параметров и без return**

````python
def say_hello():
    print("Привет, мир!")

say_hello()  # Вызов функции
````

Нет параметров.

Нет return.

Просто выполняет действие.
