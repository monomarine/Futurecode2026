### Возврат значения (return)

**Особенности `return`:**
- Завершает работу функции.
- Может возвращать любые данные (числа, строки, списки и т.д.).
- Если нет `return`, функция возвращает `None`.

```python
def add(a, b):
    return a + b

result = add(3, 5)
print(result)  # 8
```

### Множественный return

Функция может иметь несколько точек выхода:

```python
def is_even(number):
    if number % 2 == 0:
        return True
    else:
        return False

print(is_even(4))  # True
print(is_even(5))  # False
```

Упрощенный вариант:

```python
def is_even(number):
    return number % 2 == 0
```
