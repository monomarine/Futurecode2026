## Числовые типы данных

В Python есть три основных числовых типа:

| Тип | Описание | Пример |
|-----|----------|--------|
| `int` | целые числа (неограниченной длины!) | `42`, `-7`, `0`, `10**50` |
| `float` | числа с плавающей точкой (дробные) | `3.14`, `-0.5`, `2.0` |
| `complex` | комплексные числа | `3+4j` |



### Целые числа (int)

Целые числа в Python **не имеют ограничения по размеру** (в отличие от C/C++/Java, где `int` — 32 или 64 бита):

```python
big = 10 ** 100
print(big)  #
```

Это удобно, но требует больше памяти для больших чисел.

Запись целых чисел:

```python
a = 42          # десятичная
b = 0b1010      # двоичная (10)
c = 0o17        # восьмеричная (15)
d = 0xFF        # шестнадцатеричная (255)

# Разделитель разрядов для читаемости
e = 1_000_000   # то же самое, что 1000000
```

### Дробные числа (float)

`float` хранится в формате IEEE 754 (двойная точность, 64 бита).

```python
x = 3.14
y = -0.5
z = 2.0
w = 1e-3
v = 2.5e6
```

**Важная особенность — погрешность вычислений.** Не все десятичные дроби точно представимы в двоичной системе:

```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

### Операции с числами

Арифметика (мы уже видели в теме 3, повторяем кратко):

```python
print(7 + 3)      # 10
print(7 - 3)      # 4
print(7 * 3)      # 21
print(7 / 3)      # 2.3333333333333335 (всегда float)
print(7 // 3)     # 2 (целочисленное деление)
print(7 % 3)      # 1 (остаток)
print(7 ** 3)     # 343 (степень)
```

**Смешивание типов.** Если один операнд — `int`, а другой — `float`, результат будет `float`:

```python
print(2 + 3.0)     # 5.0 (float)
print(10 / 2)      # 5.0 (float)
print(10 // 2)     # 5  (int, потому что // возвращает int при int-операндах)
print(10.0 // 2)   # 5.0 (float, потому что один из операндов float)
```

### Полезные функции и модуль `math`

Встроенные функции:

```python
print(abs(-7))         # 7
print(round(3.14159, 2))  # 3.14
print(round(3.5))         # 4 (банковское округление: round(2.5) == 2!)
print(pow(2, 10))         # 1024
print(divmod(17, 5))      # (3, 2)
print(min(3, 1, 4, 1, 5)) # 1
print(max(3, 1, 4, 1, 5)) # 5
```

**Осторожно с `round()`:** Python использует «банковское округление» (округление к чётному):


```python
print(round(2.5))  # 2
print(round(3.5))  # 4
print(round(1.5))  # 2
```

### Модуль `math`:

его необходимо импотировать в начале файла с кодом чтобы получить доступ к функционалу:

```python
import math

print(math.sqrt(16))       # 4.0
print(math.pi)             # 3.141592653589793
print(math.e)              # 2.718281828459045
print(math.floor(3.7))     # 3
print(math.ceil(3.2))      # 4
print(math.trunc(-3.7))    # -3 (отбрасывает дробную часть)
print(math.gcd(12, 18))    # 6 (НОД)
print(math.factorial(5))   # 120
```
