## Строковый тип данных Str

**Строка** — это упорядоченная последовательность символов, заключённая в кавычки.

```python
s1 = 'одинарные'
s2 = "двойные"
s3 = '''тройные
на несколько строк'''
s4 = """тоже
тройные"""
```

Тип строки — `str`, и строки **неизменяемы** (immutable): нельзя изменить отдельный символ, можно только создать новую строку.

```python
s = "Python"
# s[0] = "J"   # ОШИБКА: TypeError
s = "J" + s[1:]  # создать новую строку
print(s)         # Jython
```

### Индексация и срезы

Каждый символ имеет индекс, начиная с `0`. С отрицательными индексами — с конца:

```python
s = "Python"
#     P  y  t  h  o  n
#     0  1  2  3  4  5
#    -6 -5 -4 -3 -2 -1

print(s[0])
print(s[1])
print(s[-1])
print(s[-2])
```

**Срез (slice)** — извлечение подстроки:

s[start:stop:step]

- `start` — с какого индекса (включительно), по умолчанию 0;
- `stop` — до какого индекса (не включая), по умолчанию конец строки;
- `step` — шаг, по умолчанию 1.


```python
s = "Python"

print(s[0:3])
print(s[:3])
print(s[2:])
print(s[:])
print(s[::2])
print(s[::-1])
print(s[-3:])
print(s[2:-1])
```

**Срез не выходит за границы** — это безопасно:

```python
print(s[0:100])  # 'Python' (без ошибки)
print(s[100:])   # '' (пустая строка)
```

Но **индексация за границей** — ошибка:

```python
# print(s[100])  # IndexError: string index out of range
```

### Методы строк

Строки имеют много методов.

Далее перечисленны ключевые методы (все они **возвращают новую строку**, не меняют исходную):

#### Работа с регистром

```python
s = "Hello World"
print(s.lower())
print(s.upper())
print(s.title())
print(s.capitalize())
print(s.swapcase())
```

#### Поиск и замена

```python
s = "Python — это просто"
print(s.find("это"))
print(s.find("Java"))
print(s.index("это"))
# s.index("Java")
print(s.count("о"))
print(s.replace("просто", "легко"))
```

#### Разделение и объединение:

```python
s = "яблоко, банан, апельсин"
parts = s.split(", ")
print(parts)

words = "один два три".split()
print(words)

fruits = ['яблоко', 'банан', 'апельсин']
print(", ".join(fruits))
```

**`split()` без аргумента** разбивает по любым пробельным символам (пробелы, табы, переводы строк) и **удаляет пустые элементы** — очень удобно для обработки ввода.

#### Очистка и выравнивание

```python
s = "  привет  "
print(s.strip())       # 'привет' (убирает пробелы с обоих концов)
print(s.lstrip())      # 'привет  '
print(s.rstrip())      # '  привет'

print("abc".ljust(10, '-'))  # 'abc-------'
print("abc".rjust(10, '-'))  # '-------abc'
print("abc".center(10, '-')) # '---abc----'

print("42".zfill(5))   # '00042' (дополнение нулями)
```

#### Проверки

```python
print("123".isdigit())    # содержание цифровых символов
print("abc".isalpha())    # содержание буквенных символов
print("abc123".isalnum()) # или буквы или числа
print("   ".isspace())    # пробельный символ
print("Hello".startswith("He"))  # начало с конкретных символов
print("Hello".endswith("lo"))    # окончание конкретными символами
```

### f - строки

Самый современный и удобный способ форматирования строк - перед началом строки ставится символ `f` а внутри строки в фигурных скобках {} (плесхолдерах) перечисляются переменные или выражение - которые автоматически при формировании строки будут подставлять значения

```python
name = "Анна"
age = 18
print(f"Меня зовут {name}, мне {age} лет.")

print(f"Через год мне будет {age + 1}")
print(f"Длина имени: {len(name)}")
```

#### Форматирование чисел в f-строках

есть специальные описатели которые ставятся после двоеточия в плейсхолдерах. Они управляют внешним видом отображаемых данных  - ограничивают нужным количеством символов, добавить символ процента и тд

```python
pi = 3.141592653589793

print(f"{pi:.2f}")     # 3.14 (2 знака после запятой)
print(f"{pi:.0f}")     # 3
print(f"{42:5d}")      # '   42' (ширина 5, выравнивание вправо)
print(f"{42:<5d}")     # '42   ' (влево)
print(f"{42:^5d}")     # ' 42  ' (по центру)
print(f"{1234567:,}")  # '1,234,567' (разделители тысяч)
print(f"{0.25:.1%}")   # '25.0%' (проценты)
print(f"{255:x}")      # 'ff' (шестнадцатеричное)
```
