## Логический тип данных `bool`

### Значения `True` и `False`

олько два значения: `True` и `False` (с заглавной буквы!). Это подтип `int`, где `True == 1`, `False == 0`:

```python
print(True == 1)    # True
print(False == 0)   # True
print(True + True)  # 2 (!!)
print(True * 10)    # 10
```

Это может пригодиться, но злоупотреблять не стоит — читаемость важнее.

### Логические операторы

| Оператор | Значение       | Пример                     |
| -------- | -------------- | -------------------------- |
| `and`    | логическое И   | `True and False` → `False` |
| `or`     | логическое ИЛИ | `True or False` → `True`   |
| `not`    | логическое НЕ  | `not True` → `False`       |

```python
print(True and False)
print(not True)
```

### Операторы сравнения

Возвращают `bool`:

```python
print(5 > 3)
print(5 < 3)
print(5 <= 4)
print(5 == 5)
print(5 != 5)
```

#### **Цепочки сравнений** (уникальная особенность Python):

```python
x = 5
print(0 < x < 10)      # True (эквивалентно 0 < x and x < 10)
print(0 < x > 3)       # True (0 < x and x > 3)
print(1 < 2 < 3 < 4)   # True
```

### Приведение к bool и «ложные» значения

Любое значение в Python можно использовать в логическом контексте (например, в `if`). Есть **«ложные» (falsy) значения**:

- `False`
- `0`, `0.0`
- `""` (пустая строка)
- `[]`, `()`, `{}`, `set()` (пустые коллекции)
- `None`

Все остальные значения — **«истинные» (truthy)**:

```python
print(bool(0))       # False
print(bool(42))      # True
print(bool(""))      # False
print(bool("0"))     # True (непустая строка!)
print(bool([]))      # False
print(bool([0]))     # True
print(bool(None))    # False
```

### Приоритет логических операторов

Порядок: `not` > `and` > `or`.

```python
print(True or False and False)   # True  (False and False = False, True or False = True)
print((True or False) and False) # False (скобки меняют порядок)
```
