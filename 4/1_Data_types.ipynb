## Что такое данные и тип данных

**Данные** — это информация, с которой работает программа: числа, текст, логические значения, наборы значений.

**Тип данных** — это характеристика объекта, которая определяет:
- какие значения он может принимать,
- какие операции с ним допустимы,
- сколько памяти он занимает (в Python это скрыто от программиста, но тип влияет на поведение).

В Python тип объекта можно узнать функцией `type()`:

```python
print(type(42))        # <class 'int'>
print(type(3.14))      # <class 'float'>
print(type("Привет"))  # <class 'str'>
print(type(False))      # <class 'bool'>
print(type([1, 2, 3])) # <class 'list'>
```

в Python всё является объектом, и у каждого объекта есть тип.

Python — язык с **динамической типизацией**: тип определяется значением, а не объявлением переменной. Одна и та же переменная может менять тип.

```python
x = 10          # int
x = "десять"    # теперь str
x = 10.0        # теперь float
```

### **Изменяемые (mutable) и неизменяемые (immutable) типы**

| Неизменяемые (immutable) | Изменяемые (mutable) |
| ------------------------ | -------------------- |
| `int`, `float`, `bool`   | `list`               |
| `str`                    | `dict`               |
| `tuple`                  | `set`                |

Неизменяемый объект нельзя изменить «на месте» — любая операция создаёт **новый** объект. Изменяемый можно менять, не создавая новый. Это различие будет постоянно всплывать дальше (особенно в списках и словарях).
