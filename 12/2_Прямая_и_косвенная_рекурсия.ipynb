### Прямая и косвенная рекурсия

- **Прямая рекурсия** — функция вызывает саму себя напрямую (как в примере с факториалом).
- **Косвенная рекурсия** — функция вызывает другую функцию, которая, в свою очередь, вызывает первую.

```python
def is_even(n):
    if n == 0:
        return True
    return is_odd(n - 1)

def is_odd(n):
    if n == 0:
        return False
    return is_even(n - 1)

print(is_even(10))  # True
print(is_odd(7))    # True
```
