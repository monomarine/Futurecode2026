## Что такое рекурсия

**Рекурсия** — это метод решения задач, при котором функция вызывает саму себя. Рекурсивные функции используются для решения задач, которые можно разбить на более простые подзадачи того же типа.

**Классический пример — факториал числа:**


```python
def factorial(n):
    if n == 1:          # условие выхода (базовый случай)
        return 1
    return n * factorial(n - 1)  # рекурсивный вызов

print(factorial(5))  # 120
```

**Как это работает:**
- `factorial(5)` → `5 * factorial(4)`
- `factorial(4)` → `4 * factorial(3)`
- `factorial(3)` → `3 * factorial(2)`
- `factorial(2)` → `2 * factorial(1)`
- `factorial(1)` → `1` (базовый случай, рекурсия останавливается)
- Результат: `5 * 4 * 3 * 2 * 1 = 120`



**Плюсы и минусы рекурсии:**

| Плюсы | Минусы |
| :--- | :--- |
| Позволяет элегантно решать некоторые задачи | Может быть медленнее итерации |
| Код часто короче и понятнее | Потребляет много памяти (стек вызовов) |

**Бесконечная рекурсия**

**Проблема:** Если рекурсивная функция вызывает саму себя бесконечно, это приводит к переполнению стека (`RecursionError`).

```python
def countdown(n):
    print(n)
    countdown(n - 1)  # Нет условия остановки!

countdown(5)  # Бесконечный вывод, затем ошибка RecursionError
```

### Решение: добавить базовый случай (условие выхода)

```python
def countdown(n):
    if n <= 0:          # Условие выхода
        print("Пуск!")
        return
    print(n)
    countdown(n - 1)

countdown(5)
```
