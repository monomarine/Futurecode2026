```python
file = open('test.txt') # открытие ресурса файла
data = file.read() # чтение из файла
print(data) # вывод содержимого
file.close() # закрытие ресурса файла
```

```python
file = open('test.txt','w')
for i in range(5):
    text = input("введите текст для записи в файл") + "\n"
    file.writelines(text)

file.close()
```

```python
file = open('test.txt','r')
lines = file.readlines()

print(lines)
file.close()
```

```python
file = open('test.txt','r')

while True:
    line = file.readline()
    if line == "stop\n":
        break
    print(line)

file.close()
```
