## Построение и анализ дерева рекурсивных вызовов в Python

**Рекурсивная функция** — функция, которая в своём теле содержит явные или косвенные вызовы самой себя.

- **Явная рекурсия** — функция вызывает себя напрямую (по имени).
- **Косвенная рекурсия** — функция вызывает другую функцию, которая в итоге снова вызывает первую.

**Бесконечная рекурсия** возникает, когда нет условия выхода

### Дерево рекурсивных вызовов

**Дерево рекурсивных вызовов** — графическая модель, визуализирующая процесс выполнения рекурсивной функции.


Каждый **узел** дерева — один конкретный вызов функции со своим набором аргументов.

````python
def factorial(n):
    if n == 1:              # условие выхода (базовый случай)
        return 1
    return n * factorial(n - 1)  # рекурсивный вызов

print(factorial(5))  # 120
````

Дерево вызовов для `factorial(5)` — это цепочка (линейное дерево):

```
factorial(5)
└── factorial(4)
    └── factorial(3)
        └── factorial(2)
            └── factorial(1)  → возвращает 1
```

Если в коде **несколько рекурсивных вызовов**, дерево становится ветвящимся, и его размер растёт очень быстро (вычисление чисел Фибоначчи)

### Глубина рекурсии и лимит вызовов

**Глубина рекурсии** — длина самого длинного пути от корня дерева до листа. Она равна максимальному количеству одновременно существующих в памяти кадров стека.

Управление лимитом — через модуль `sys`:

````python
import sys

print("Текущий лимит:", sys.getrecursionlimit())  # обычно 1000

sys.setrecursionlimit(3000)                        # установка нового лимита
print("Текущий лимит:", sys.getrecursionlimit())  # 3000
````

становка слишком высокого значения лимита может привести к немедленному аварийному завершению программы (переполнение стека / segfault).

### Кэширование вызовов функции

**Кэширование** запоминает результаты вызовов функции. При повторном вызове с теми же аргументами результат возвращается из памяти (кэша) без повторного выполнения кода.

**Без кэширования** функция выполняется каждый раз:

````python
def slow_sum(a, b):
    print("Выполняем тяжелый расчет...")
    return a + b

print(slow_sum(2, 3))  # текст + 5
print(slow_sum(2, 3))  # снова текст + 5
````

**С кэшированием** (декоратор `@cache` из `functools`) второй вызов берёт ответ из кэша:

````python
from functools import cache

@cache
def slow_sum(a, b):
    print("Выполняем тяжелый расчет...")
    return a + b

print(slow_sum(2, 3))  # текст + 5
print(slow_sum(2, 3))  # только 5 (текста нет)
````
