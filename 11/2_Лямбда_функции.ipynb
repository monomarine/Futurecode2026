## Lambda-функции

**Lambda-функции** — это анонимные (безымянные) функции, которые можно создать с помощью ключевого слова `lambda`. Они компактнее обычных функций, ограничены одним выражением и не требуют придумывать им имена.

**Синтаксис:**
```python
lambda аргументы: выражение
```


````python
# Возведение в квадрат
square = lambda x: x ** 2
print(square(5))  # 25
````

````python
# Сумма двух чисел
sum_numbers = lambda a, b: a + b
print(sum_numbers(8, 10))  # 7
print(sum_numbers("8", "10"))  # 810
````

### Примеры применения Lambda-функций

Арифметические действия

````python
double = lambda x: x * 2
print(double(5))  # 10
````

Фильтрация

````python
numbers = [1, 2, 3, 4, 5, -3, -5, -212]
f = lambda x : not x % 2 == 0 # лямбда для проверки на нечетность
f2 = lambda y : y < 0
f3 = lambda x : x > 0
even = list(filter(lambda x : x % 2 == 0, numbers))
print(even)  # [2, 4]
````

Сортировка

````python
nums = [3, 1, 4, 2]
sorted_nums = sorted(nums, key= lambda x : -x)
print(sorted_nums)  # [4, 3, 2, 1]
````

**Используйте `lambda` только для простых однострочных операций. Для сложной логики — `def`.**

сортивка строковых элементо

````python
names = ['яков','константин','федор', 'алек']

sort_key = lambda s : -len(s)
sorted_names = sorted(names, key=sort_key)
print(sorted_names)
````
