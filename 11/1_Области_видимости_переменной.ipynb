**Область видимости (scope)** переменной в Python — это часть программы, в которой переменная доступна для использования. Она определяет, где и как переменная может быть использована в коде.

В Python переменные могут иметь разные области видимости:
- **Локальные** — создаются внутри функций и доступны только внутри этих функций.
- **Глобальные** — создаются в основном теле программы и доступны во всех частях программы.

### Локальные переменные

Существуют только внутри функции. Если попытаться обратиться к ним снаружи — будет ошибка.

```python
def func():
    x = 10  # локальная переменная
    print(x)

func()      # 10
#print(x)    # Ошибка: NameError: name 'x' is not defined
```

### Глобальные переменные

Доступны во всей программе.

```python
y = 20  # глобальная переменная

def func():
    print(y)  # 20

func()
```

### Ключевое слово `global`

Если нужно изменить глобальную переменную внутри функции, используется ключевое слово `global`.

```python
y = 20  # глобальная переменная

def func():
    global y
    y = 30  # изменяем глобальную переменную
    print(y)

func()      # 30
print(y)    # 30
```
