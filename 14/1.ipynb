## Структуры данных в Python

**Структура данных** — способ организации и хранения данных, который определяет, какие операции с ними можно выполнять и насколько эффективно.

В Python есть четыре основные встроенные структуры: `list`, `tuple`, `dict`, `set`. Выбор правильной структуры напрямую влияет на скорость работы программы и читаемость кода.

### Списки (`list`)

**Когда использовать:**
- нужна изменяемая последовательность элементов;
- важен порядок элементов;
- требуется частая модификация (добавление/удаление);
- реализация стека или очереди.

**Примеры задач:** обработка последовательности с изменением, хранение истории операций, сортировка и фильтрация данных.

```python
nums = [5, 2, 8, 1]
nums.append(3)        # добавление в конец
nums.sort()           # сортировка на месте
print(nums)           # [1, 2, 3, 5, 8]
```

### Кортежи (`tuple`)

**Когда использовать:**
- данные не должны изменяться;
- нужно гарантировать неизменность;
- кортеж используется как ключ словаря (список — нельзя);
- нужно вернуть несколько значений из функции;
- важна производительность (кортежи быстрее и легче списков).

**Примеры задач:** конфигурационные параметры, координаты точек, записи из БД.

```python
point = (10, 20)
x, y = point          # распаковка
print(x, y)           # 10 20

config = {(1920, 1080): "FHD", (3840, 2160): "4K"}  # кортеж как ключ
```

### Словари (`dict`)

**Когда использовать:**
- данные в формате «ключ — значение»;
- нужен быстрый поиск по ключу (O(1));
- подсчёт частоты элементов;
- ассоциативные массивы;
- вложенные структуры данных.

**Примеры задач:** телефонная книга, кэширование, подсчёт слов в тексте, хранение параметров объектов.

```python
phone = {"Иван": "+7-900-111", "Сергей": "+7-900-222"}
print(phone["Иван"])          # быстрый доступ по ключу
print(phone.get("Светлана", "нет")) # безопасный доступ с значением по умолчанию
```

### Множества (`set`)

**Когда использовать:**
- нужна коллекция уникальных элементов;
- операции над множествами (объединение, пересечение, разность);
- быстрая проверка принадлежности (O(1));
- порядок не важен.

**Примеры задач:** удаление дубликатов, поиск общих друзей, проверка правописания, фильтрация уникальных посетителей.

```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a & b)   # пересечение: {3}
print(a | b)   # объединение: {1, 2, 3, 4, 5}
print(3 in a)  # True — проверка наличия
```
