## 4. Операции с числами

### Арифметические операторы:

```python
a = 10
b = 3

print(a + b)   # 13  — сложение
print(a - b)   # 7   — вычитание
print(a * b)   # 30  — умножение
print(a / b)   # 3.3333333333333335 — деление (всегда float!)
```

> оператор `/` **всегда** возвращает `float`, даже если делится нацело:

```python
print(10 / 2)   # 5.0 (float, не int)
print(type(10 / 2))  # <class 'float'>
```

### Целочисленное деление `//` и остаток `%`
**Целочисленное деление** `//` отбрасывает дробную часть (округляет вниз к минус бесконечности):

```python
print(10 // 3)   # 3
print(10 // 2)   # 5
print(7 // 2)    # 3
print(-7 // 2)   # -4 (округление вниз!)
```

**Остаток от деления** `%`:

```python
print(10 % 3)    # 1
print(10 % 2)    # 0
print(7 % 2)     # 1
print(-7 % 2)    # 1 (в Python остаток всегда неотрицательный при положительном делителе)
```

### Возведение в степень `**`

```python
print(2 ** 3)     # 8
print(2 ** 10)    # 1024
print(9 ** 0.5)   # 3.0 (квадратный корень)
print(2 ** -1)    # 0.5
```

`**` работает и с дробными показателями — это удобный способ извлечь корень.

### Приоритет операций и скобки
Порядок выполнения (от высшего к низшему):

1. `**` (возведение в степень)
2. `*`, `/`, `//`, `%`
3. `+`, `-`
4. Операторы сравнения (`<`, `>`, ` ==`, ...)
5. `not`, `and`, `or` (логические)

```python
print(2 + 3 * 4)      # 14, а не 20
print((2 + 3) * 4)    # 20
print(2 ** 3 ** 2)    # 512, потому что ** правоассоциативен: 2**(3**2)
print(10 - 3 - 2)     # 5, потому что - левоассоциативен: (10-3)-2
```

если сомневаешься — ставь скобки. Скобки повышают читаемость и исключают ошибки.

### Полезные встроенные функции

```python
print(abs(-5))       # 5 — модуль числа
print(abs(3.14))     # 3.14

print(round(3.14159, 2))  # 3.14 — округление до 2 знаков
print(round(3.7))         # 4 — до целого

print(min(3, 7, 1, 9))    # 1 — минимум
print(max(3, 7, 1, 9))    # 9 — максимум

print(divmod(10, 3))      # (3, 1) — частное и остаток одной функцией
```

`divmod(a, b)` возвращает кортеж `(a // b, a % b)` — удобно, когда нужны оба результата.
