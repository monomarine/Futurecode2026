# Переменные

**Переменная** — это имя, которое ссылается на значение, хранящееся в памяти компьютера. Через переменную мы можем обращаться к этому значению и изменять его.

```python
age = 18
name = "Анна"
pi = 3.14159
is_student = True
```

После выполнения этих строк:
- `age` ссылается на `18`
- `name` ссылается на `"Анна"`
- `pi` ссылается на `3.14159`
- `is_student` ссылается на `True`

> переменная не «хранит» значение в буквальном смысле — она ссылается на объект в памяти компьютера. В Python имя — это ссылка на объект.

##### как Python связывает имя и объект
В Python переменная — это **ссылка** на объект в оперативной памяти компьютера. Можно убедиться в этом с помощью `id()` — она возвращает адрес объекта в памяти:

```python
a = 100
b = a
print(id(a))
print(id(b))
```

Обе переменные ссылаются на **один и тот же объект** `100`.

А теперь:

```python
a = 100
b = 100
print(id(a) == id(b))
```

### Требования к именованию переменных
В Python есть **жёсткие правила** (нарушение → `SyntaxError`) и **соглашения** (нарушение → не ошибка, но плохой стиль).


#### Жёсткие правила
1. **Имя может содержать** буквы (a–z, A–Z), цифры (0–9) и символ подчёркивания `_`.

```python
user_name = "Анна"   # OK
userName = "Анна"    # OK (но не по стилю Python)
name2 = "Борис"      # OK
```

2. **Имя не может начинаться с цифры.**

```python
2name = "Ошибка"     # SyntaxError
name2 = "OK"         # OK
```

3. **Имя не может содержать пробелы и спецсимволы.**

```python
my name = "Ошибка"   # SyntaxError
my-name = "Ошибка"   # SyntaxError
my_name = "OK"
```

4. **Имя не может совпадать с зарезервированным словом** (keyword).

```python
if = 5      # SyntaxError — if это ключевое слово
for = 10    # SyntaxError
class = 1   # SyntaxError
```

5. **Регистрозависимость.** `age`, `Age` и `AGE` — три **разные** переменные.

```python
age = 18
Age = 20
print(age)  # 18
print(Age)  # 20
```

#### Соглашения PEP 8
PEP 8 — официальный стиль кода Python. Для переменных рекомендуется:
- **`snake_case`** — слова в нижнем регистре, разделённые подчёркиванием:

```python
user_name = "Анна"       # хорошо
userName = "Анна"        # плохо (стиль Java/C#)
UserName = "Анна"        # плохо (стиль классов)
```

- **Осмысленные имена.**

```python
a = 18                 # плохо: непонятно, что это
student_age = 18       # хорошо
```

- **Не использовать имена, похожие на встроенные функции:**

```python
list = [1, 2, 3]       # плохо — перекрывает встроенную list()
my_list = [1, 2, 3]    # хорошо
```

### Множественное присваивание и обмен значений
Python позволяет присваивать значения нескольким переменным в одной строке:

```python
a, b = 10, 20
print(a, b)  # 10 20
```

А также **менять значения местами** без временной переменной:

```python
a = 1
b = 2

a, b = b, a
print(a, b)  # 2 1
```

### Динамическая типизация

Python — **динамически типизированный** язык: тип переменной определяется значением, которое на неё ссылается в данный момент. Одна и та же переменная может менять тип.

```python
x = 10          # x ссылается на int
print(type(x))  # <class 'int'>

x = "Привет"    # теперь x ссылается на str
print(type(x))  # <class 'str'>

x = 3.14        # теперь float
print(type(x))  # <class 'float'>
```

Это не ошибка — это нормальное поведение Python. Но злоупотреблять этим не стоит: переменная с «плавающим» типом ухудшает читаемость.
