## ввод-вывод в консоли

**Консоль** (или терминал) - это текстовое окно, через которое пользователь общается с программой: программа что-то выводит на экран, а пользователь может вводить данные с клавиатуры.

почти любая программа строится по схеме **«вход → обработка → выход»**.
Консоль - самый простой способ организовать вход и выход.



### Консольный вывод

`print()` выводит значения в консоль.
Это одна из самых используемых функций в Python.

`print()` автоматически переводит курсор на новую строку после каждого вызова.

```python
print("Привет, мир!")
print(42)
print(3.14)
print(True)
```

#### Вывод нескольких значений
при передаче нескольких аргументов в print() через запятую - значения выведутся разделенные пробелами по умолчанию

```python
print("Возраст:", 18)
print("Имя:", "Анна", "Возраст:", 18)
```

**`sep` (separator)** задаёт строку-разделитель между аргументами.

```python
print("Имя", "Анна", "Возраст", 18, sep=", ")
print("2024", "01", "15", sep="-")
```

#### управление окончанием строки вывода
По умолчанию `end="\n"` (перевод строки).

Если задать `end=""` или `end=" "`, вывод не будет переходить на новую строку.

```python
print("Первая строка", end=" ")
print("продолжение на той же строке")
```

#### Экранированные символы

Внутри строк можно использовать специальные последовательности:

| Последовательность | Значение                                   |
| ------------------ | ------------------------------------------ |
| `\n`               | перевод строки                             |
| `\t`               | табуляция                                  |
| `\\`               | обратный слэш                              |
| `\"`               | кавычка внутри строки в двойных кавычках   |
| `\'`               | кавычка внутри строки в одинарных кавычках |

```python
print("Строка 1\nСтрока 2")
print("Колонка1\tКолонка2")
print("Он сказал: \"Привет!\"")
```

#### Формат строки (f-строки)
удобный способ вставлять значения переменных прямо в текст.

Перед строкой ставится буква `f`, а внутри в фигурных скобках `{}` указываются переменные или выражения.

```python
name = "Анна"
age = 18
print(f"Меня зовут {name}, мне {age} лет.")
```

### Консольный ввод

 `input()` читает строку, которую пользователь вводит с клавиатуры, и возвращает её.

```python
name = input("Введите ваше имя: ")
print("Привет,", name)
```

`input()`всегда возвращает строку (`str`), даже если пользователь ввёл число.

Если пользователь ввёл `5`, переменная `x` содержит строку `"5"`, а не число `5`.

```python
x = input("Введите число: ")
print(type(x))
```

Строка внутри `input()` — это приглашение к вводу, которое видит пользователь. Она выводится без перевода строки.

```python
age = input("Сколько вам лет? ")
```

возможна ошибка - сложение строки и числа

```python
x = input("Введите число: ")
y = x + 1   # ОШИБКА!
```

### Преобразование введённых данных
Чтобы работать с введёнными числами, нужно явно преобразовать строку в число или в другой нужный тип данных:


```python
# Целое число
a = int(input("Введите целое число: "))

# Дробное число
b = float(input("Введите дробное число: "))
```

если пользователь введёт `"abc"`, `int()` выбросит ошибку `ValueError`. Обработку ошибок (try/except) будем изучать позже, но знать о ней полезно.
