## 3. Оператор присваивания

Общий вид присваивания значений в переменную:

имя_переменной = выражение


Справа (выражение) может быть:
- литерал: `x = 5`
- другая переменная: `x = y`
- арифметическое выражение: `x = a + b * 2`
- вызов функции: `x = int(input())`
- результат любой операции: `x = (a > b)`

```python
x = 5
y = x + 10
z = y * 2
name = input("Имя: ")
```

### Отличие присваивания от сравнения

| Оператор | Назначение                                        | Пример   |
| -------- | ------------------------------------------------- | -------- |
| `=`      | **присваивание** (положить значение в переменную) | `x = 5`  |
| `==`     | **сравнение** (проверить, равны ли значения)      | `x == 5` |

```python
x = 5          # присваивание: x становится 5
print(x == 5)  # сравнение: True
print(x == 6)  # сравнение: False
```

Внутри `if` используется именно ` ==`:

```python
x = 5
if x == 5:
    print("x равно пяти")
```

> Если случайно написать `if x = 5:` — будет синтаксическая ошибка, потому что в условии присваивание недопустимо (без специальных конструкций).

### Составные операторы присваивания

Python поддерживает сокращённую запись «вычислить и присвоить»:

| Оператор | Эквивалент | Пример |
|----------|-----------|--------|
| `+=` | `x = x + y` | `x += 3` |
| `-=` | `x = x - y` | `x -= 2` |
| `*=` | `x = x * y` | `x *= 4` |
| `/=` | `x = x / y` | `x /= 2` |
| `//=`| `x = x // y`| `x //= 3` |
| `%=` | `x = x % y` | `x %= 5` |
| `**=`| `x = x ** y`| `x **= 2` |

```python
x = 10
x += 5     # x = 15
x -= 3     # x = 12
x *= 2     # x = 24
x //= 5    # x = 4
x **= 2    # x = 16
print(x)   # 16
```

Составные операторы — это **синтаксический сахар**: они не меняют семантику, но делают код короче и читаемее

### Порядок вычисления: сначала справа, потом присваивание

При выполнении `x = выражение` Python сначала **полностью вычисляет выражение справа**, и только потом присваивает результат переменной.

```python
x = 5
x = x + 1
```

Пошагово:
1. Вычисляется `x + 1` → `5 + 1` → `6` (старое значение `x` = 5)
2. Результат `6` присваивается `x`
3. Теперь `x` = 6

Это важно понимать: в правой части используется **старое** значение `x`, а не новое.

Аналогично:

```python
a = 10
a = a * a   # a = 10 * 10 = 100
print(a)    # 100
```

### Цепочка присваиваний
Можно присвоить одно значение сразу нескольким переменным:

```python
a = b = c = 0
print(a, b, c)  # 0 0 0
```

Здесь все три переменные ссылаются на один и тот же объект `0`. Для неизменяемых объектов (числа, строки) это безопасно.

Можно и так:

```python
x, y, z = 1, 2, 3
print(x, y, z)  # 1 2 3
```

Это называется **распаковкой** — количество переменных слева должно совпадать с количеством значений справа (выше уже рассматривался пример)
