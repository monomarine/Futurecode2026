## Функция `range`

`range()` — встроенная функция, которая **порождает последовательность чисел**.

Она не создаёт список в памяти (ленивая), а выдаёт числа по одному. Это экономит память.

**Три формы:**

```python
range(stop)              # 0, 1, 2, ..., stop-1
range(start, stop)       # start, start+1, ..., stop-1
range(start, stop, step) # start, start+step, ... до stop (не включая)
```

**Ключевое правило:** `stop` **не включается** в последовательность.

### Форма `range(stop)`

````python
for i in range(20):
    print(i, end=' ')
# 0 1 2 3 4
````

`range(5)` → `0, 1, 2, 3, 4` (5 чисел, начиная с 0).

````python
# Повторить действие 10 раз
for _ in range(10):
    print("Привет")
````

### Форма `range(start, stop)`

````python
for i in range(10, 20):
    print(i, end=' ')
# 2 3 4 5 6
````

`range(2, 7)` → `2, 3, 4, 5, 6` (7 **не включается**).

### Форма `range(start, stop, step)`

````python
for i in range(10, 1000, 100):
    print(i, end=' ')
````

````python
for i in range(10, 0, -1):
    print(i, end=' ')
````

````python
for i in range(5, 0, -1):
    print(i)
````

**`step` может быть отрицательным** — тогда последовательность идёт вниз. Для этого `start > stop`.

**`step` не может быть 0:**

### `range` не создаёт список

`range` — это «ленивый» объект: числа вычисляются по мере необходимости. Это важно для больших диапазонов:

````python
r = range(5)
print(r)          # range(0, 5)
print(list(r))    # [0, 1, 2, 3, 4]
print(len(r))     # 5
print(r[0])       # 0
print(r[-1])      # 4
print(3 in r)     # True
````

````python
# Не занимает гигабайты памяти
r = range(1_000_000_000)
print(len(r))     # 1000000000
print()
````
