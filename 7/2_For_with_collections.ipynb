### `for` по строке

Строка — последовательность символов. `for` перебирает её посимвольно:

```python
for char in "Python":
    print(char)
```

**Пример: подсчёт гласных**

```python
text = "я изучаю питон аааааааааааааа"
vowels = "аеёиоуыэюяАЕЁИОУЫЭЮЯ"
count = 0
for ch in text:
    if ch in vowels:
        count += 1
print(f"Гласных: {count}")
```

### `for` по списку и кортежу

```python
fruits = ["яблоко", "апельсин", "мандарин"]
for fruit in fruits:
    print(fruit)
```

```python
point = (3, 5)
for coord in point:
    print(coord)
```

**Пример: сумма элементов списка**

```python
numbers = [10, 20, 30]
total = 0
for number in numbers:
    total += number
print(f"Сумма: {total}")   # 100
```

### `for` по словарю

По словарю можно итерироваться по ключам, значениям или парам:

```python
student = {"name": "Анна", "age": 18, "city": "Москва"}

# По ключам (по умолчанию)
for key in student:
    print(key)
```

```python
# По ключам явно
for key in student.keys():
    print(key)
```

```python
# По значениям
for value in student.values():
    print(value)
```

```python
# По парам (ключ, значение)
for key, value in student.items():
    print(f"{key:5} : {value}")
```

### `for` по множеству

порядок обхода множества **не гарантирован**

```python
colors = {"red", "green", "blue", "blue"}
for color in colors:
    print(color)
```
