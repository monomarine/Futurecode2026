# Цикл с параметром

Цикл for — это цикл, который проходит по элементам последовательности (или повторяет действие заданное число раз). В отличие от while, где мы явно проверяем условие, for автоматически перебирает элементы и сам останавливается, когда они заканчиваются

````python
# while: вручную управляем счётчиком i
i = 0
while i < 5:
    print(i)
    i += 1

# for: Python сам перебирает элементы
for i in range(5):
    print(i)
````

Оба дают 0 1 2 3 4, но for короче, безопаснее и читаемее:

- не нужно помнить про инициализацию счётчика;
- не нужно помнить про его изменение (нет риска бесконечного цикла);
- не нужно вручную проверять условие окончания.

## отличия While и For

| `while`                              | `for`                                        |
| ------------------------------------ | -------------------------------------------- |
| Работает, **пока условие истинно**   | Проходит **по элементам последовательности** |
| Число итераций может быть неизвестно | Число итераций = длина последовательности    |
| Легко получить бесконечный цикл      | Бесконечного цикла «по ошибке» не бывает     |
| Нужен ручной счётчик                 | Счётчик автоматический                       |

### Когда использовать for:
- Пройти по всем элементам списка, строки, кортежа, словаря, множества.
- Повторить действие N раз (for i in range(N)).
- Обработать последовательность чисел с шагом.
- Пройти по индексам (for i in range(len(lst))).

### Когда использовать while:
- Число итераций заранее неизвестно (ввод до условия, поиск, игра).
- Условие выхода сложное и не сводится к «перебрать все элементы».

### Базовый синтаксис цикла с параметром

```python
for переменная in последовательность:
    # тело цикла (с отступом)
    действие
```

**Как работает:**

1. Берётся первый элемент из `последовательности` и присваивается `переменной`.
2. Выполняется тело цикла.
3. Берётся следующий элемент → снова тело.
4. Когда элементы закончились — цикл завершается.

**Синтаксические правила (как у `while` и `if`):**

- После `for ... in ...` — **двоеточие `:`**.
- Тело — с **отступом** (4 пробела).

````python
for i in range(5):
    print(i)
````

**Пошаговый разбор:**

| Итерация | `i` | Действие |
|----------|-----|----------|
| 1 | 0 | print(0) |
| 2 | 1 | print(1) |
| 3 | 2 | print(2) |
| 4 | 3 | print(3) |
| 5 | 4 | print(4) |
| — | — | элементы кончились → выход |

`range(5)` порождает последовательность `0, 1, 2, 3, 4` — **5 чисел, начиная с 0**.


### Имя переменной цикла

Переменная цикла — обычная переменная. Её имя должно отражать смысл элементов:

````python
# Плохо но иногда ОК
for x in [1, 2, 3]:
    print(x)

# Хорошо
for number in [1, 2, 3]:
    print(number)

# Если элемент не используется — принято имя _
for _ in range(3):
    print("Привет")
````

 после цикла переменная сохраняет **последнее** значение:

````python
for i in range(5):
    pass
print(i)   # 4 — последнее значение
````

Если последовательность пуста — цикл не выполнится ни разу, и переменная цикла **не будет создана**:

````python
for i in range(0):
    pass
# print(i)  # NameError: name 'i' is not defined
````
