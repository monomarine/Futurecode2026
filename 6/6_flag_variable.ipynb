## Переменная-флаг

**Переменная-флаг** (флажок) — это переменная `bool`, которая **сигнализирует о каком-то событии** во время цикла. До цикла флаг ставится в одно состояние (обычно `False`), внутри — может измениться, после цикла — используется для принятия решения.

```python
flag = False          # до цикла — событие не произошло

while ...:
    if условие_события:
        flag = True   # событие произошло
        # (иногда break, иногда нет)

if flag:
    # обрабатываем, если событие было
else:
    # обрабатываем, если не было
```

### Пример: поиск элемента

````python
numbers = [3, 7, 2, 9, 5]
target = 9

found = False   # переменная - флаг
i = 0
while i < len(numbers):
    if numbers[i] == target:
        found = True
        break
    i += 1

if found:
    print(f"Нашли {target} на позиции {i}")
else:
    print(f"{target} не найден")
````

Здесь `found` — флаг: «нашли ли». После цикла он определяет, что делать.

### Пример: проверка, все ли числа положительные

````python
numbers = [3, 7, -2, 9, 5]

all_positive = True     # предполагаем, что все положительные
i = 0
while i < len(numbers):
    if numbers[i] <= 0:
        all_positive = False
        break
    i += 1

if all_positive:
    print("Все положительные")
else:
    print("Есть неположительные")
````

**Логика флага:** «предполагаем лучшее, ищем опровержение». Как только нашли нарушение — флаг `False`, `break`.

### Флаг-счетчик

Иногда флаг — не `bool`, а **счётчик** совпадений:

````python
numbers = [3, 7, 3, 9, 3, 5]
target = 3

count = 0 # флаг-счетчик
i = 0
while i < len(numbers):
    if numbers[i] == target:
        count += 1
    i += 1

print(f"Число {target} встречается {count} раз")
````
