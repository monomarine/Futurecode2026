## Синтаксис While цикла

```python
while условие:
    # тело цикла (с отступом)
    действие
```

**Как работает:**

1. Проверяется `условие`.
2. Если **истинно** — выполняется тело цикла, затем снова шаг 1.
3. Если **ложно** — цикл завершается, управление переходит к коду после цикла.

**Синтаксические правила (как у `if`):**

- После `while` — условие, затем **двоеточие `:`**.
- Тело — с **отступом** (4 пробела).
- Тело может содержать несколько инструкций.

````python
i = 0
while i < 5:
    print(i)
    i += 1
print("Цикл завершён")
````

**Пошаговый разбор:**

| Шаг | `i` | `i < 5`? | Действие |
|-----|-----|----------|----------|
| 1 | 0 | True | print(0), i = 1 |
| 2 | 1 | True | print(1), i = 2 |
| 3 | 2 | True | print(2), i = 3 |
| 4 | 3 | True | print(3), i = 4 |
| 5 | 4 | True | print(4), i = 5 |
| 6 | 5 | False | выход из цикла |

**Критически важные детали:**

1. **Переменная-счётчик инициализируется ДО цикла.** Если забыть `i = 0`, будет `NameError`.
2. **В теле должно быть изменение условия.** Здесь `i += 1` — без него `i` всегда 0, условие всегда истинно → **бесконечный цикл**.
3. **Порядок действий в теле важен.** Если написать `i += 1` перед `print(i)`, вывод будет `1 2 3 4 5`.

````python
i = 0
while i < 5:
    i += 1
    print(i)
````

### Обратный отсчёт через цикл

````python
i = 5
while i > 0:
    print(i)
    i -= 1
print("Старт!")
````

### Накопление суммы

````python
total = 0
i = 1
while i <= 100:
    total += i
    i += 1
print(f"Сумма 1..100 = {total}")   # 5050
````

**Схема «накопителя»:**

- До цикла: `total = 0` (начальное значение).
- В теле: `total += ...` (прибавляем к накопленному).
- После цикла: `total` содержит результат.

Тот же приём работает для произведения (`prod = 1; prod *= i`), для подсчёта количества (`count += 1`), для поиска максимума (`if x > mx: mx = x`).

### Ввод до правильного значения

Классический случай, где `while` незаменим — **число повторений заранее неизвестно**:

````python
password = ""
while password != "secret":
    password = input("Введите пароль: ")
print("Доступ разрешён")
````

Программа будет запрашивать пароль, пока не введут `"secret"`. Сколько итераций будет — неизвестно.

### Проверка корректности ввода

````python
n = int(input("Введите положительное число: "))
while n <= 0:
    print("Ошибка! Число должно быть положительным.")
    n = int(input("Попробуйте ещё раз: "))
print(f"Вы ввели: {n}")
````

### Обработка цифр числа

Этот приём — «разбор числа по цифрам» — очень частый в задачах.

````python
n = 12345
while n > 0:
    digit = n % 10        # последняя цифра
    print(digit, end=" ")
    n //= 10              # отбрасываем последнюю цифру
````
