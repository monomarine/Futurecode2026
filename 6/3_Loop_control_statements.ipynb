## Управление циклом: `continue` и `break`

Иногда нужно **досрочно прервать** цикл или **пропустить** часть итерации. Для этого есть два оператора.

### `break` — досрочный выход из цикла

`break` **немедленно завершает** цикл, даже если условие ещё истинно. Управление переходит к первой инструкции после цикла.

```python
i = 0
while True:
    if i == 3:
        break
    print(i)
    i += 1
print("После цикла")
```

на `i = 3` выполняется `break` → цикл прерывается, `print(3)` не выполняется, сразу идёт `print("После цикла")`.

#### Алгоритм поиска перебором

```python
numbers = [3, 7, 2, 9, 5, 8]
target = 9
i = 0
found = False
while i < len(numbers):
    if numbers[i] == target:
        found = True
        break
    i += 1

if found:
    print(f"Найдено на позиции {i}")
else:
    print("Не найдено")
```

Как только нашли — дальше искать не нужно, `break` экономит время.

**`break` выходит только из ОДНОГО (ближайшего) цикла.** Если циклы вложенные — выйдет только из внутреннего.

### `continue` — пропуск остатка итерации

`continue` **немедленно переходит к следующей итерации**, пропуская оставшиеся инструкции тела. Условие цикла проверяется заново.

```python
i = 0
while i < 5:
    i += 1
    if i == 3:
        continue
    print(i)
```

на `i = 3` выполняется `continue` → `print(3)` пропускается, цикл переходит к проверке условия и следующей итерации.

#### Алгоритм фильтрации

```python
i = 0
while i < 10:
    i += 1
    if i % 2 == 0:
        continue        # пропускаем чётные
    print(i)            # печатаем только нечётные
```

### Разница между `break` и `continue`

| Оператор   | Что делает                                                     |
| ---------- | -------------------------------------------------------------- |
| `break`    | **полностью завершает** цикл                                   |
| `continue` | **пропускает остаток текущей итерации**, переходит к следующей |
