# Циклы

Цикл** — это конструкция, которая позволяет **повторять** один и тот же блок кода несколько раз.

Без циклов программисту пришлось бы писать одну и ту же инструкцию многократно:

````python
# Плохо: 5 раз вручную
print("Привет")
print("Привет")
print("Привет")
print("Привет")
print("Привет")
````

````python
# Хорошо: один раз написали — 5 раз выполнилось
for i in range(50):
    print("Привет")
````

**Зачем нужны циклы:**

1. **Повторение действия** известное число раз.
2. **Обработка коллекций** (пройти по всем элементам списка, строки, словаря).
3. **Работа до выполнения условия** (пока пользователь не введёт правильные данные; пока не закончится файл; пока не достигнут результат).
4. **Накопление результата** (сумма, произведение, счётчик, поиск максимума).

**Два основных вида циклов в Python:**

| Цикл | Когда использовать |
|------|-------------------|
| `while` | когда **число повторений заранее неизвестно**, зависит от условия |
| `for` | когда нужно **пройти по последовательности** или повторить **известное число раз**  |


**Схема работы любого цикла:**

```
проверить условие → если истинно: выполнить тело → снова проверить условие → ...
                  → если ложно: выйти из цикла
```

**итерация.** Одна итерация = одно выполнение тела цикла.
