## Бесконечный цикл. Остановка бесконечного цикла

**Бесконечный цикл** — цикл, условие которого **всегда истинно**, поэтому он никогда не завершается сам.

```python
while True:
    print("Это никогда не закончится")
```

### Когда бесконечный цикл полезен

Сознательный `while True` с `break` — нормальный приём, когда **число итераций заранее неизвестно**, а условие выхода удобнее проверить **внутри** цикла:

```python
while True:
    command = input("Команда (exit — выход): ")
    if command == "exit":
        break
    print(f"Выполняю: {command}")
print("Программа завершена")
```

### Остановка бесконечного цикла: `break`

Основной способ выйти из `while True` — **`break`**:

```python
while True:
    n = int(input("Введите положительное число (0 — выход): "))
    if n == 0:
        break
    if n < 0:
        print("Отрицательное, попробуйте ещё")
        continue
    print(f"Вы ввели {n}")
```
